```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

# Travel domain dataset (user queries and bot responses)
data = [
   ("Hi", "Hello! Welcome to the Travel Assistant. Where would you like to go?"),
    ("Suggest me a travel destination", "How about visiting Paris? It's famous for the Eiffel Tower and rich culture."),
    ("Tell me about accommodations", "You can find hotels, hostels, and Airbnb options depending on your budget."),
    ("What transportation options are available?", "You can travel by flights, trains, or buses. For city tours, taxis and rental cars are great."),
    ("Recommend an activity", "If you're in Dubai, try a desert safari. In Switzerland, explore the scenic train rides."),
    ("Bye", "Goodbye! Have a safe journey!")
]

# Separate data into inputs and responses
inputs, responses = zip(*data)

# Create a text classification pipeline
model = Pipeline([
    ('vectorizer', CountVectorizer()),
    ('classifier', MultinomialNB())
])
# Train the model
model.fit(inputs, responses)
# Function to interact with the chatbot
def chat_with_bot():
    print("Type 'Bye' to end the chat.")
    while True:
        user_input = input("You: ")
        if user_input.lower() == 'bye':
            print("Chatbot: Goodbye!")
            break
        else:
            response = model.predict([user_input])[0]
            print(f"Chatbot: {response}")
# Start chatting with the bot
chat_with_bot()
```
